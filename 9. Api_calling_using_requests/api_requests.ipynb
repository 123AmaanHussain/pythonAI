```python
res = requests.get(
    "https://api.frankfurter.app/latest",
    params={"from": "USD", "to": "EUR"}
)
data = res.json()
print("1 USD =", data['rates']['EUR'], "EUR")
```

```python
try:
    res = requests.get(
    "https://api.frankfurter.app/latest?from=USD&to=INR"
    )
    if res.status_code == 200:
        data = res.json()
        print("1 USD =", data['rates']['INR'], "INR")
    else:
        print("Something went wrong. Status Code:", res.status_code)

except Exception as e:
    print("Couldn't reach api:", e)  
```

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

res = requests.get("https://api.frankfurter.app/latest?from=USD&to=INR")
rate = res.json()['rates']['INR']

prompt = f'''
    I have 200 NVIDIA stocks, today the price is 215$.
    How much money will I get in INR if I sell them all today.
    Today's USD to INR conversion rate is {rate}
'''
try:
    interaction = client.interactions.create(
        model="gemini-4-hydrogen",
        input=prompt
    ).output_text
except Exception as e:
    interaction = f"Something went wrong: {e}"

print(interaction)
```
