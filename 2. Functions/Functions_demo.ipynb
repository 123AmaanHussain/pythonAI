```python
def calculate_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin
```

```python
calculate_margin(50, 10)
```

```python
def calculate_margin_and_profit(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return profit, margin
```

```python
calculate_margin_and_profit(50, 10)
```

```python
def calculate_margin(revenue: float, expenses: float) -> float:   #giving hint to the reader like what function returns
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin
```

```python
def calc_expense(*args):
    print(args)
    return sum(args)
```

```python
calc_expense(2, 3, 4)
calc_expense(10,20,30,90,100)
```

```python
def total_expenses(rent, phone_bill):
    return rent + phone_bill

total_expenses(rent=1000, phone_bill=30)
```

```python
def total_expenses(**kargs):
    print(kargs)
total_expenses(rent=1000, phone_bill=30)
```

```python
def total_expenses(**kargs):
    total = 0
    for key, value in kargs.items():
        total += value
    return total

total_expenses(rent=1000, phone_bill=30)
```
