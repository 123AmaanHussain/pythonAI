```python
revenue = 0
expenses = 10

try:
    margin = (revenue - expenses) * 100 / revenue
    print(f"{margin:.2f}")
except ZeroDivisionError:
    print("Cannot Divide by zero")
```

```python
def safe_margin(revenue, expenses):
    try:
        return (revenue - expenses) * 100 / revenue
    except ZeroDivisionError:
        return None

print(safe_margin(50, 10))
print(safe_margin(0, 100))
```
