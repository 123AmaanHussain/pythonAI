```python
with open("joke.txt", "w") as f:            #context manager no need to f.close() later it will handle
    f.write("My memory is so bad.\n")
    f.write("How bad it is?.\n")
    f.write("How bad is what?.\n")
```

```python
with open("joke.txt", "r") as f:
    #print(f.readlines())
   # for line in f:
      #  print(line)
    document = f.read()
print(document)
```

```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

prompt = f"""
    Here is the document I shared: 
    {document}

    Summarize it in 3 bullet points
"""
```

```python
try:
    interaction = client.interactions.create(
        model="gemini-4-hydrogen",
        input=prompt
    )
    summary = interaction.output_text
except Exception as e:
    summary = f"Something went wrong: {e}"

print(summary)
```

```python
with open("summary.txt","w", encoding="utf-8") as f:
    f.write(summary)
print("Summary saved to txt file")
```

```python
import json

recipe = {
    "title" : "Yellow Dal",
    "calories" : 280,
    "ingredients" : ["lentils", "onion", "tomato", "cumin"]
}

with open("recipe.json","w", encoding="utf-8") as f:
    json.dump(recipe, f, indent=2)

print("recipe saved")
```
