```python
revenues = [50, 60, 55]
expenses = [10, 30, 35]
```

```python
#list iteration
for rev in revenues:
    print(f"Revenues: {rev}")
```

```python
for i in range(len(revenues)):
    print(revenues[i])
```

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(revenues[i])
    print(expenses[i])
    print(f"Margin {i + 1}: {margin : .2f}")
    print(f"Profit {i + 1}: {profit}")
```

```python
margins = []

for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    margins.append(margin)

margins
```

### Dictionary

```python
fin = { "revenue" : 50, 
        "expenses" : 10 
    }

type(fin)
```

```python
profit = fin["revenue"] - fin["expenses"]
profit
```

```python
fin['profit'] = fin['revenue'] - fin['expenses']  #create new key-value pair
fin
```

```python
financials = [
    { "revenue" : 50, 
        "expenses" : 10 
    },
    { "revenue" : 60, 
        "expenses" : 30 
    },
    { "revenue" : 55, 
        "expenses" : 35 
    }
]
financials
```

```python
for fin in financials:
    profit = fin["revenue"] - fin['expenses']
    fin['margin'] = profit * 100 / fin['revenue']
financials
```

```python
financials = {
    "Q1" : {"revenue" : 50, "expenses" : 10},
    "Q2" : {"revenue" : 60, "expenses" : 30},
    "Q3" : {"revenue" : 55, "expenses" : 35},
}

for quarter, data in financials.items():
    print("Quarter: ", quarter)
    print("data: ", data)
```

```python
financials = {
    "Q1" : {"revenue" : 50, "expenses" : 10},
    "Q2" : {"revenue" : 60, "expenses" : 30},
    "Q3" : {"revenue" : 55, "expenses" : 35},
}

for quarter, data in financials.items():
    margin = (data['revenue'] - data['expenses']) / data['revenue'] * 100
    print(f"{quarter} : {margin:.2f}%")
```

```python
financials = {
    "Q1" : {"revenue" : 50, "expenses" : 10},
    "Q2" : {"revenue" : 60, "expenses" : 30},
    "Q3" : {"revenue" : -55, "expenses" : 35},
    "Q4" : {"revenue" : 10, "expenses" : 35}             #error handlings
}

for quarter, data in financials.items():
    if data['revenue'] < 0 or data['expenses'] < 0:
        print(f"Invalid data for {quarter}....... Skippinggg.....")
        continue
    margin = (data['revenue'] - data['expenses']) / data['revenue'] * 100
    print(f"{quarter} : {margin:.2f}%")
```
