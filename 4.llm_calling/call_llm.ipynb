```python
from google import genai

client = genai.Client()

interaction = client.interactions.create(
    model="gemini-4-hydrogen",
    input="Explain how AI works in a few words"
)
print(interaction.output_text)
```

```python
ingradients = ["dal", "onion", "tomatoes"]

','.join(ingradients)
```

```python
def generate_recipe(ingradients : list, cuisine, diet):
    '''
    Generate a recipe using ingradients provided.
    Cuisine can be "Italian", "Mexican", "Indian" etc.
    diet can be "vegan", "vegetarian", "Jain" etc.
    '''

    prompt = f'''
    Generate one food receipe using these ingradients: {','.join(ingradients)}
    Receipe should not be more than 100 words
    Cuisine: {cuisine}
    Diet: {diet}
    '''

    interaction = client.interactions.create(
        model="gemini-4-hydrogen",
        input=prompt
    )
    return interaction.output_text
```

```python
recipe = generate_recipe(["Tomatoes", "lentils", "onion","jeera","salt"], "Indian", "Vegan")
recipe
```
