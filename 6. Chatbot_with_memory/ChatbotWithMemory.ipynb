```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
```

```python
conversations = []

def chat(user_message):
    conversations.append({"role": "user","parts": [{"text": user_message}]})

    try:
        response = client.models.generate_content(
            model="gemini-4-hydrogen",
            contents = conversations
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    conversations.append({"role": "model", "parts":[{"text": reply}]})
    return reply
```

```python
chat("Hi! My name is Amaan. Please remember it.")
```

```python
chat("Hi! whats my name")
```

```python
for message in conversations:
    print(message["role"], "->", message["parts"][0]["text"])
    print()
```
